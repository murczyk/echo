# Regenerate keywords to highlight

This script generates a list of keywords in all of its forms to put into keywords.rs

```python
keywords = ["птичка", "небо", "птица", "глаза", "ворона", "курятник", "летать", "летает"]
```

```python
import spacy
import pymorphy3
import json

nlp = spacy.load("ru_core_news_md")
morph = pymorphy3.MorphAnalyzer()

lexemes = []
for keyword in keywords:
    lexemes.extend([x.word for x in morph.parse(keyword)[0].lexeme])
print(json.dumps(lexemes, ensure_ascii=False, indent=2))
```
